## NPS Relacional consolidado por clientes

```sql
CREATE OR REPLACE TABLE data_lake_hermes_ai.prata.nps_relacional_consolidado (
SELECT COUNT(*) AS Total_tickets,
Cod_Cliente,
ROUND(AVG(resposta_NPS),0) AS media_resposta_NPS,
CASE 
    WHEN (ROUND(AVG(resposta_NPS),0)) <= 6 THEN 'detrator'
    WHEN (ROUND(AVG(resposta_NPS),0)) > 6 AND (ROUND(AVG(resposta_NPS),0)) <= 8 THEN 'passivo'
    WHEN (ROUND(AVG(resposta_NPS),0)) >= 9 THEN 'promotor'
    ELSE 'sem_classificacao'
END AS grupo_nps,
ROUND(AVG(resposta_unidade),0) AS media_resposta_unidade,
ROUND(AVG(Nota_SupTec_Agilidade),0) AS media_SupTec_Agilidade,
ROUND(AVG(Nota_SupTec_Atendimento),0) AS media_SupTec_Atendimento,
ROUND(AVG(Nota_Comercial),0) AS media_comercial,
ROUND(AVG(Nota_Custos),0) AS media_custos,
ROUND(AVG(Nota_AdmFin_Atendimento),0) AS media_AdmFin_Atendimento,
ROUND(AVG(Nota_Software),0) AS media_software,
ROUND(AVG(Nota_Software_Atualizacao),0) AS media_Software_Atualizacao
FROM (SELECT 
    respondedAt,
    metadata_codcliente AS Cod_Cliente,
    COALESCE(resposta_NPS, 0) AS resposta_NPS,
    COALESCE(TRY_CAST(resposta_unidade AS INT), 0) AS resposta_unidade,
    COALESCE(TRY_CAST(Nota_SupTec_Agilidade AS INT), 0) AS Nota_SupTec_Agilidade,
    COALESCE(TRY_CAST(Nota_SupTec_Atendimento AS INT), 0) AS Nota_SupTec_Atendimento,
    COALESCE(TRY_CAST(Nota_Comercial AS INT), 0) AS Nota_Comercial,
    COALESCE(TRY_CAST(Nota_Custos AS INT), 0) AS Nota_Custos,
    COALESCE(TRY_CAST(Nota_AdmFin_Atendimento AS INT), 0) AS Nota_AdmFin_Atendimento,
    COALESCE(TRY_CAST(Nota_Software AS INT), 0) AS Nota_Software,
    COALESCE(TRY_CAST(Nota_Software_Atualizacao AS INT), 0) AS Nota_Software_Atualizacao
FROM data_lake_hermes_ai.bronze.nps_relacional)
GROUP BY 2)
```
