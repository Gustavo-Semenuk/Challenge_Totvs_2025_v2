# Validation Totvs (Dados dos Clientes)

```sql

CREATE OR REPLACE TABLE validation_totvs AS (
SELECT
a.CLIENTE,
m.MRR_12M,
b.DS_PROD,
b.DS_LIN_REC,
b.CIDADE,
b.DS_CNAE,
b.DS_SEGMENTO,
b.DS_SUBSEGMENTO,
b.FAT_FAIXA,
b.MARCA_TOTVS,
b.MODAL_COMERC,
b.PAIS,
b.PERIODICIDADE,
b.SITUACAO_CONTRATO,
b.UF,
b.VL_TOTAL_CONTRATO,
b.DT_ASSINATURA_CONTRATO,
a.CLIENTE_DESDE,
date_diff(now(),a.CLIENTE_DESDE) as Tempo_Conta
FROM totvs.default.clientes_desde a 
LEFT JOIN totvs.default.dados_clientes b ON a.CLIENTE=b.CD_CLIENTE
LEFT JOIN totvs.default.mrr m ON m.CLIENTE=a.CD_CLIENTE)

```

```sql
SELECT
DISTINCT UF,
CIDADE
FROM validation_totvs
```
